##Python Control Structure

```python
numbers = [12,75,130,180,145,525,50]
for number in numbers:
    if number>150 and number<=500:
        print(number)
    elif number > 500:
        y=number**2
        print (y)


```

##R Control structure


```python
%%R

numbers = c(12,75,130,180,145,525,50)

```

```python
%%R

for (number in numbers) {
    if (number > 150 && number <= 500) {
        print(number)
    } else if (number > 500) {
        y = number^2
        print(y)
    }
}
```
